```python
// =====================================================
// FIGURE 1: STUDY AREA MAP (Google Earth Engine - JavaScript)
// =====================================================

// 1. Study area
var addisAbaba = ee.Geometry.Point([38.75, 9.02]);
var aoi = addisAbaba.buffer(15000);       // 15 km buffer
var wide = addisAbaba.buffer(65000);      // For context

// 2. Sentinel-2 tile footprint (T37PDK)
var s2 = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
    .filterBounds(aoi)
    .filterDate('2021-01-01', '2021-12-31');

var s2Tile = s2.first();
var tileFootprint = ee.Feature(s2Tile.geometry());
var tileID = s2Tile.get('MGRS_TILE');
print('Sentinel-2 Tile:', tileID);

// 3. ESA WorldCover land cover
var worldCover = ee.ImageCollection('ESA/WorldCover/v200').first();

var worldCoverPalette = [
  '006400','ffbb22','ffff4c','f096ff','fa0000',
  'b4b4b4','f0f0f0','0064c8','0096a0','00cf75','fae6a0'
];

Map.addLayer(worldCover.clip(wide),
  {min: 10, max: 100, palette: worldCoverPalette, opacity: 0.85},
  'Land Cover');

// 4. Sentinel-2 tile outline (blue)
Map.addLayer(
  ee.Image().byte().paint({
    featureCollection: ee.FeatureCollection([tileFootprint]),
    color: 1, width: 3
  }),
  {palette: ['0000FF']},
  'Tile T37PDK'
);

// 5. AOI 15 km buffer (red)
Map.addLayer(
  ee.Image().byte().paint({
    featureCollection: ee.FeatureCollection([ee.Feature(aoi)]),
    color: 1, width: 3
  }),
  {palette: ['FF0000']},
  'AOI (15 km)'
);

// 6. Addis Ababa marker (yellow)
Map.addLayer(ee.Feature(addisAbaba), {color: 'yellow'}, 'Addis Ababa');

// 7. Legend
var legend = ui.Panel({
  style: {position: 'bottom-left', padding: '8px',
          backgroundColor: 'white', border: '1px solid black', width: '180px'}
});
legend.add(ui.Label({
  value: 'Legend', style: {fontWeight: 'bold', fontSize: '13px', margin: '0 0 6px 0'}
}));

var legendItems = [
  {label: 'AOI (15 km)', color: 'FF0000'},
  {label: 'Tile 37PDK', color: '0000FF'},
  {label: 'Addis Ababa', color: 'FFFF00'},
  {label: 'Tree Cover', color: '006400'},
  {label: 'Shrubland', color: 'ffbb22'},
  {label: 'Grassland', color: 'ffff4c'},
  {label: 'Cropland', color: 'f096ff'},
  {label: 'Built-up', color: 'fa0000'},
  {label: 'Bare/Sparse', color: 'b4b4b4'},
  {label: 'Water', color: '0064c8'}
];

legendItems.forEach(function(item) {
  var cb = ui.Label({style: {backgroundColor: item.color, padding: '6px',
                             margin: '0 0 3px 0', border: '1px solid black'}});
  var ds = ui.Label({value: item.label,
                     style: {margin: '0 0 3px 6px', fontSize: '10px'}});
  legend.add(ui.Panel({widgets: [cb, ds]}));
});
Map.add(legend);

// 8. Center map
Map.centerObject(wide, 9);

print('✅ Figure 1 rendered. Take a screenshot or export.');
```
